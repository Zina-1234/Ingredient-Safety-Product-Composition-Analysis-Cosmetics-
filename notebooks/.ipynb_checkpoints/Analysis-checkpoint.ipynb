### Cleaning Strategy

```python
# Create a clean CMR column: cmr_clean
data["cmr_clean"] = (
    data["CMR"]
    .astype(str)                     # convert NaN to string temporarily
    .replace("nan", pd.NA)           # back to proper NaN
    .str.replace("equael", "equal", regex=False)  # fix the typo
    .str.replace(r"\(\)", "", regex=True)         # remove empty parentheses
    .str.replace(r"\\n", " ", regex=True)        # replace literal \n with space
    .str.replace(r"\s+", " ", regex=True)        # collapse multiple spaces
    .str.strip()                                # remove leading/trailing spaces
)
```

```python
# Create a clean Name of Common Ingredients Glossary column
# 1. Clean the original column
data["common_ingredients_clean"] = (
    data["Name of Common Ingredients Glossary"]
    .astype("string")                               # proper nullable string dtype
    .str.strip()                                   # remove leading/trailing whitespace
    .str.replace(r"\s*/\s*", "; ", regex=True)     # unify '/' as ';'
    .str.replace(r"\s*;\s*", "; ", regex=True)     # clean spacing around ';'
    .str.replace(r"\s+", " ", regex=True)          # collapse multiple spaces
    .str.title()                                   # normalize case
)

# 2. Fallback: if glossary is missing, use Chemical name / INN
data["common_ingredients_final"] = data["common_ingredients_clean"]

mask = data["common_ingredients_final"].isna()

data.loc[mask, "common_ingredients_final"] = (
    data.loc[mask, "Chemical name / INN"]
        .astype("string")
        .str.strip()
        .str.title()
)

# 3. Split into list
data["common_ingredients_list"] = (
    data["common_ingredients_final"]
    .str.split(r";\s*")
)

# 4. Explode into one ingredient per row
df_exploded = data.explode("common_ingredients_list").reset_index(drop=True)

# 5. Final ingredient cleaning
df_exploded["ingredient"] = (
    df_exploded["common_ingredients_list"]
    .str.strip()
    .str.upper()
)

# Optional safety filter (should remove almost nothing now)
df_exploded = df_exploded[df_exploded["ingredient"].notna()]
```

```python
# remove administrative junk
df_exploded = df_exploded[
    df_exploded["ingredient"] != "MOVED OR DELETED"
]
```

```python
# Identify duplicated ingredients (including all occurrences)
duplicated_ingredients = df_exploded[
    df_exploded.duplicated(subset=["ingredient"], keep=False)
]

duplicated_ingredients.sort_values("ingredient")
```

```python
# Drop dupliated ingredients
ingredient_reference = (
    df_exploded
    .drop_duplicates(subset=["ingredient"], keep="first")
    .reset_index(drop=True)
)
```

```python
# sanity check
ingredient_reference["ingredient"].duplicated().sum()
```

```python
import re
# Recreate product_type_clean explicitly
df_exploded["product_type_clean"] = (
    df_exploded["Product Type, body parts"]
    .astype(str)
    .replace("nan", pd.NA)
    .str.replace(r"\n", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)


df_analysis = df_exploded[["ingredient", "cmr_clean", "product_type_clean"]].copy()

def clean_product_type(text):
    if pd.isna(text):
        return pd.NA
    text = re.sub(r"\n", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    
    # Split ONLY on true regulatory separators
    parts = re.split(r"\(\w\)|;|\n", text)
    
    return [p.strip().title() for p in parts if p.strip()]

df_analysis["product_type"] = df_analysis["product_type_clean"].apply(clean_product_type)

df_analysis = df_analysis.explode("product_type").reset_index(drop=True)

df_analysis = df_analysis.drop_duplicates(
    subset=["ingredient", "product_type"]
).reset_index(drop=True)

df_analysis = df_analysis[["ingredient", "cmr_clean", "product_type"]]
```

### Cleaning Strategy

```python
# Tools & Brushes are not cosmetic products
data4["primary_category"].unique()
```

```python
# Keep columns
cols_to_keep = [
    "product_id", "product_name", "brand_name",
    "ingredients", "price_usd", "rating", "reviews", "loves_count",
    "primary_category", "secondary_category", "tertiary_category",
    "sephora_exclusive", "online_only", "limited_edition", "new"
]

df3_clean = data4[cols_to_keep].copy()
```

```python
# Drop missing ingredients
df3_clean = df3_clean[df3_clean["ingredients"].notna()]
```

```python
# keep only cosmetic categories: drop non cosmetic categories
cosmetic_categories = [
    "Fragrance", "Bath & Body", "Hair",
    "Makeup", "Skincare", "Men", "Mini Size", "Gifts"
]

df3_clean = df3_clean[
    df3_clean["primary_category"].isin(cosmetic_categories)
]
```

```python
df3_clean = df3_clean[df3_clean["primary_category"] != "Tools & Brushes"]
```

```python
df3_clean["primary_category"] = (
    df3_clean["primary_category"]
    .str.upper()
    .str.strip()
)
```

```python
# Clean ingredient column
import ast

def parse_ingredient_list(x):
    if pd.isna(x):
        return []
    if isinstance(x, list):
        return x
    try:
        return ast.literal_eval(x)
    except:
        return []
```

```python
df3_clean["ingredients_parsed"] = df3_clean["ingredients"].apply(parse_ingredient_list)
```

```python
type(df3_clean.loc[0, "ingredients_parsed"]) == list
```

```python
def split_ingredients(lst):
    out = []
    for item in lst:
        parts = item.split(",")
        out.extend(parts)
    return out
```

```python
df3_clean["ingredients_flat"] = df3_clean["ingredients_parsed"].apply(split_ingredients)
```

```python
import re

def clean_inci(x):
    if pd.isna(x):
        return pd.NA
    x = x.upper()
    x = re.sub(r"\(.*?\)", "", x)   # remove parentheses
    x = re.sub(r"[^A-Z\s]", "", x)  # remove dots, brackets, quotes
    x = re.sub(r"\s+", " ", x).strip()
    return x
```

```python
df3_exploded = df3_clean.explode("ingredients_flat").reset_index(drop=True)
df3_exploded["ingredient"] = df3_exploded["ingredients_flat"].apply(clean_inci)
```

```python
df3_exploded = df3_exploded[df3_exploded["ingredient"].notna()]
```

```python
df3_exploded.head(2)
```

## Data Preparation & Merging

We have three datasets (cosing, sephora and skincare), our goal is to create a relational structure (One table = one entity):
- ingredients table → ingredient + CMR info
- products table → product info
- product_ingredients table → linking table (Relationship): product × ingredient × product type

### Ingredients Table: ingredients_regulatory (COSING)

```python
# Extract unique ingredients from cosing
ingredients_df = (df_analysis[["ingredient", "cmr_clean"]].drop_duplicates().reset_index(drop=True))

# Add surrogate key for SQL
ingredients_df["ingredient_id"] = ingredients_df.index + 1 # simple unique ID

# Rename, reorder columns for SQL clarity
ingredients_df = ingredients_df.rename(columns={"cmr_clean": "cmr"})
ingredients_df = ingredients_df[["ingredient_id", "ingredient", "cmr"]] 

ingredients_df.head(2)
```

### # Regulatory usage table

```python
ingredient_usage_df = (
    df_analysis[["ingredient", "product_type"]].merge(ingredients_df[["ingredient_id","ingredient"]], on="ingredient", how="left"))
ingredient_usage_df = ingredient_usage_df[["ingredient_id", "product_type"]].drop_duplicates().reset_index(drop=True)
```

```python
ingredient_usage_df.head(2)
```

### Products Table (Sephora: main data)

```python
# Select relevant product info from sephora dataframe (product identity)
sephora_products = df3_exploded[[
    "product_id", "product_name", "brand_name", "price_usd", "rating", "reviews",
    "loves_count", "primary_category", "secondary_category", "tertiary_category",
    "sephora_exclusive", "online_only", "limited_edition", "new"
]].drop_duplicates(subset="product_id").reset_index(drop=True)

sephora_products.head(2)
```

```python
## CHECK !!!!!!
# Map to ingredient_id
sephora_ingredients = df3_exploded.merge(
    ingredients_df[["ingredient_id","ingredient"]],
    on="ingredient",
    how="left"
)
sephora_ingredients = sephora_ingredients[["product_id", "ingredient_id"]].drop_duplicates().reset_index(drop=True)
```

### Product table (Skincare dataset)

```python
# Skincare products table
skincare_products = df1_exploded[[
    "product_id", "product_name", "brand_name", "price_usd", "rating",
    "reviews", "loves_count", "primary_category", "secondary_category",
    "tertiary_category", "sephora_exclusive", "online_only",
    "limited_edition", "new"
]].drop_duplicates().reset_index(drop=True)
```

```python
skincare_ingredients = df1_exploded.merge(
    ingredients_df[["ingredient_id","ingredient"]],
    on="ingredient",
    how="left"
)
skincare_ingredients = skincare_ingredients[["product_id", "ingredient_id"]].drop_duplicates().reset_index(drop=True)
```

#### Merge Sephora + Skincare if needed

```python
products_df = pd.concat([sephora_products, skincare_products]).drop_duplicates(subset="product_id").reset_index(drop=True)
product_ingredients_df = pd.concat([sephora_ingredients, skincare_ingredients]).drop_duplicates().reset_index(drop=True)
```

### Product × Ingredient Table

```python
##### LOOOOOOOOOOOST!!!!!!!!!!!!!!!!!!!

# Merge sephora exploded ingredient table with ingredient IDs
product_ingredients_df = df3_exploded[["product_id", "ingredient", "product_type_clean"]].copy()

# Uppercase ingredient for matching
product_ingredients_df["ingredient"] = product_ingredients_df["ingredient"].str.upper()

# Merge with ingredients_df to get ingredient_id
product_ingredients_df = product_ingredients_df.merge(
    ingredients_df,
    how="left",
    on="ingredient"
)

# Keep relevant columns and rename for SQL
product_ingredients_df = product_ingredients_df[["product_id", "ingredient_id", "ingredient", "product_type"]]
product_ingredients_df = product_ingredients_df.rename(columns={"product_type_clean": "product_type"})

# Drop duplicates
product_ingredients_df = product_ingredients_df.drop_duplicates().reset_index(drop=True)

product_ingredients_df.head(10)
```

### Save to SQL (MySQL)

```python
from sqlalchemy import create_engine

# Example connection string: adjust username/password/dbname
engine = create_engine("mysql+pymysql://username:password@localhost:3306/cosmetics_db")

# Save tables
ingredients_df.to_sql("ingredients", con=engine, if_exists="replace", index=False)
products_df.to_sql("products", con=engine, if_exists="replace", index=False)
product_ingredients_df.to_sql("product_ingredients", con=engine, if_exists="replace", index=False)
```
