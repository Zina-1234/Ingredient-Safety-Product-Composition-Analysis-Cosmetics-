### Cosing: Regulatory Data (annex III)

```python
# Cosing — Regulatory Reference
cosing_raw_0 = pd.read_excel("/Users/ZINA/Desktop/IRONHACK/Week_8/first/COSING_Annex_III_v2.xls")
```

```python
cosing_raw = cosing_raw_0.copy()
```

```python
# Create a clean CMR column: cmr (cleaned column)
cosing_raw["cmr"] = (
    cosing_raw["CMR"]
    .astype(str)                     # convert NaN to string temporarily
    .replace("nan", pd.NA)           # back to proper NaN
    .str.replace("equael", "equal", regex=False)  # fix the typo
    .str.replace(r"\(\)", "", regex=True)         # remove empty parentheses
    .str.replace(r"\\n", " ", regex=True)        # replace literal \n with space
    .str.replace(r"\s+", " ", regex=True)        # collapse multiple spaces
    .str.strip()                                # remove leading/trailing spaces
)
```

```python
# Create a clean Name of Common Ingredients Glossary column: common_ingredients (cleaned column)
# 1. Clean the original column
cosing_raw["common_ingredients"] = (
    cosing_raw["Name of Common Ingredients Glossary"]
    .astype("string")                               # proper nullable string dtype
    .str.strip()                                   # remove leading/trailing whitespace
    .str.replace(r"\s*/\s*", "; ", regex=True)     # unify '/' as ';'
    .str.replace(r"\s*;\s*", "; ", regex=True)     # clean spacing around ';'
    .str.replace(r"\s+", " ", regex=True)          # collapse multiple spaces
    .str.title()                                   # normalize case
)
```

```python
# 2. Fallback: if glossary is missing, use Chemical name / INN
mask = cosing_raw["common_ingredients"].isna()

cosing_raw.loc[mask, "common_ingredients"] = (
    cosing_raw.loc[mask, "Chemical name / INN"]
        .astype("string")
        .str.strip()
        .str.title()
)
```

```python
# 3. Split into list
cosing_raw["common_ingredients"] = (
    cosing_raw["common_ingredients"]
    .str.split(r";\s*")
)
```

```python
# 4. Explode into one ingredient per row
cosing_exploded = cosing_raw.explode("common_ingredients").reset_index(drop=True)
```

```python
# 5. Final ingredient cleaning
cosing_exploded["ingredient"] = (
    cosing_exploded["common_ingredients"]
    .str.strip()
    .str.upper()
)

# Safety filter (should remove almost nothing now)
cosing_exploded = cosing_exploded[cosing_exploded["ingredient"].notna()]
```

```python
# remove administrative junk
cosing_exploded = cosing_exploded[
    cosing_exploded["ingredient"] != "MOVED OR DELETED"
]
```

```python
import re
# Recreate product_type_clean explicitly: product_type (cleaned column)
cosing_exploded["product_type"] = (
    cosing_exploded["Product Type, body parts"]
    .astype(str)
    .replace("nan", pd.NA)
    .str.replace(r"\n", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

cosing = cosing_exploded[["ingredient", "cmr", "product_type"]].copy()
```

```python
def clean_product_type(text):
    if pd.isna(text):
        return pd.NA
    text = re.sub(r"\n", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    
    # Split ONLY on true regulatory separators
    parts = re.split(r"\(\w\)|;|\n", text)
    
    return [p.strip().title() for p in parts if p.strip()]

cosing["product_type"] = cosing["product_type"].apply(clean_product_type)

cosing = cosing.explode("product_type").reset_index(drop=True)

cosing = cosing.drop_duplicates(
    subset=["ingredient", "product_type"]
).reset_index(drop=True)

cosing = cosing[["ingredient", "cmr", "product_type"]]
```

### Sephora: Data Analysis 1

```python
# SEPHORA Catalog
sephora_raw_0 = pd.read_csv("/Users/ZINA/Desktop/IRONHACK\Week_8/first/product_info.csv/product_info.csv")
```

```python
# Tools & Brushes are not cosmetic products
sephora_raw_0["primary_category"].unique()
```

```python
# Keep columns
cols_to_keep = [
    "product_id", "product_name", "brand_name",
    "ingredients", "price_usd", "rating", "reviews", "loves_count",
    "primary_category", "secondary_category", "tertiary_category",
    "sephora_exclusive", "online_only", "limited_edition", "new"
]

sephora_raw = sephora_raw_0[cols_to_keep].copy()
```

```python
# Drop missing ingredients
sephora_raw = sephora_raw[sephora_raw["ingredients"].notna()]
```

```python
# keep only cosmetic categories: drop non cosmetic categories
cosmetic_categories = [
    "Fragrance", "Bath & Body", "Hair",
    "Makeup", "Skincare", "Men", "Mini Size", "Gifts"
]

sephora_raw = sephora_raw[
    sephora_raw["primary_category"].isin(cosmetic_categories)
]
```

```python
sephora_raw = sephora_raw[sephora_raw["primary_category"] != "Tools & Brushes"]
```

```python
# Use primary_category or cleaned product_type for product_type
sephora_raw["product_type"] = sephora_raw["primary_category"].str.title().str.strip()
```

```python
# Categories to exclude: 
exclude_categories = [
    'Brushes & Applicators', 
    'Candles & Home Scents', 
    'Beauty Supplements', 
    'Tools'
]
# Keep everything else
sephora_raw = sephora_raw[~sephora_raw["secondary_category"].isin(exclude_categories)]
```

```python
# Clean ingredient column
import ast

def parse_ingredient_list(x):
    if pd.isna(x):
        return []
    if isinstance(x, list):
        return x
    try:
        return ast.literal_eval(x)
    except:
        return []
```

```python
sephora_raw["ingredients_parsed"] = sephora_raw["ingredients"].apply(parse_ingredient_list)
```

```python
type(sephora_raw.loc[0, "ingredients_parsed"]) == list
```

```python
def split_ingredients(lst):
    out = []
    for item in lst:
        parts = item.split(",")
        out.extend(parts)
    return out
```

```python
sephora_raw["ingredients_flat"] = sephora_raw["ingredients_parsed"].apply(split_ingredients)
```

```python
import re

def clean_inci(x):
    if pd.isna(x):
        return pd.NA
    x = x.upper()
    x = re.sub(r"\(.*?\)", "", x)   # remove parentheses
    x = re.sub(r"[^A-Z\s]", "", x)  # remove dots, brackets, quotes
    x = re.sub(r"\s+", " ", x).strip()
    return x
```

```python
sephora = sephora_raw.explode("ingredients_flat").reset_index(drop=True)
sephora["ingredient"] = sephora["ingredients_flat"].apply(clean_inci)
```

```python
sephora = sephora[sephora["ingredient"].notna()]
```

```python
# Keep only relevant columns for merging with cosing
sephora = sephora[["product_id", "product_name", "brand_name", "product_type", "ingredient",
                   "price_usd", "rating", "reviews", "loves_count"]]
```

### Feature Engineering

#### Ingredient-Level Feature

```python
# CMR Flag – whether the ingredient is 
# Binary flag for CMR
ingredients_df["is_cmr"] = ingredients_df["cmr"].notna().astype(int)
ingredients_df.head()
```

```python
# Number of Product Types – how many product types the ingredient appears in
ingredient_product_types = product_ingredient_df.groupby("ingredient")["product_type"].nunique().reset_index()
ingredient_product_types.rename(columns={"product_type": "num_product_types"}, inplace=True)

ingredients_df = ingredients_df.merge(ingredient_product_types, on="ingredient", how="left")
ingredients_df["num_product_types"] = ingredients_df["num_product_types"].fillna(0).astype(int)
```

```python
# Frequency Across Products – how often each ingredient appears
ingredient_freq = product_ingredient_df["ingredient"].value_counts().reset_index()
ingredient_freq.columns = ["ingredient", "frequency_in_products"]

ingredients_df = ingredients_df.merge(ingredient_freq, on="ingredient", how="left")
ingredients_df["frequency_in_products"] = ingredients_df["frequency_in_products"].fillna(0).astype(int)
```

### Product-Level Features

```python
# Number of Ingredients
ingredients_per_product = product_ingredient_df.groupby("product_id")["ingredient"].count().reset_index()
ingredients_per_product.rename(columns={"ingredient": "num_ingredients"}, inplace=True)
```

```python
# Number of CMR Ingredients
cmr_per_product = product_ingredient_df.merge(ingredients_df[["ingredient","is_cmr"]], on="ingredient")
cmr_per_product = cmr_per_product.groupby("product_id")["is_cmr"].sum().reset_index()
cmr_per_product.rename(columns={"is_cmr": "num_cmr_ingredients"}, inplace=True)
```

```python
# CMR Ratio – fraction of ingredients that are CMR
product_features = ingredients_per_product.merge(cmr_per_product, on="product_id")
product_features["cmr_ratio"] = product_features["num_cmr_ingredients"] / product_features["num_ingredients"]
```

```python
# Optional: Average Ingredient Frequency – gives idea of “common vs rare” ingredients
ingredient_freq_map = ingredients_df.set_index("ingredient")["frequency_in_products"].to_dict()
product_ingredient_df["ingredient_freq"] = product_ingredient_df["ingredient"].map(ingredient_freq_map)

avg_freq_per_product = product_ingredient_df.groupby("product_id")["ingredient_freq"].mean().reset_index()
avg_freq_per_product.rename(columns={"ingredient_freq": "avg_ingredient_freq"}, inplace=True)

product_features = product_features.merge(avg_freq_per_product, on="product_id")
```

### Product Metadata Features

```python
# Merge product-level features with Sephora / Skincare product metadata:
# Example with Sephora
sephora_product_features = sephora_products.merge(product_features, on="product_id", how="left")

# Fill missing values for products without ingredients
sephora_product_features[["num_ingredients", "num_cmr_ingredients", "cmr_ratio", "avg_ingredient_freq"]] = \
    sephora_product_features[["num_ingredients", "num_cmr_ingredients", "cmr_ratio", "avg_ingredient_freq"]].fillna(0)
```

### Ingredient × Product Type Features

```python
# We can create aggregated features per product type:
product_type_cmr = product_ingredient_df.merge(ingredients_df[["ingredient","is_cmr"]], on="ingredient")
product_type_summary = product_type_cmr.groupby("product_type")["is_cmr"].agg(["count", "sum"]).reset_index()
product_type_summary.rename(columns={"count": "total_ingredients", "sum": "num_cmr_ingredients"}, inplace=True)
product_type_summary["cmr_ratio"] = product_type_summary["num_cmr_ingredients"] / product_type_summary["total_ingredients"]
product_type_summary.head()
```

## SQL Schema + Analytical Queries

### PREPARE DATAFRAMES

```python
# dim_product (from Sephora): One row per product
dim_product = (
    sephora_clean
    .groupby("product_id")
    .agg(
        product_name=("product_name", "first"),
        brand_name=("brand_name", "first"),
        primary_category=("product_type", "first"),
        price_usd=("price_usd", "first"),
        rating=("rating", "first"),
        loves_count=("loves_count", "first"),
        num_ingredients=("ingredient", "count")
    )
    .reset_index()
)
```

```python
# Add CMR features
# Join regulatory info
sephora_cmr = sephora_clean.merge(
    cosing_clean[["ingredient", "cmr"]],
    on="ingredient",
    how="left"
)

cmr_stats = (
    sephora_cmr
    .groupby("product_id")
    .agg(
        num_cmr_ingredients=("cmr", lambda x: x.notna().sum())
    )
    .reset_index()
)

dim_product = dim_product.merge(cmr_stats, on="product_id", how="left")
dim_product["num_cmr_ingredients"] = dim_product["num_cmr_ingredients"].fillna(0)

dim_product["has_cmr"] = (dim_product["num_cmr_ingredients"] > 0).astype(int)
dim_product["cmr_ratio"] = dim_product["num_cmr_ingredients"] / dim_product["num_ingredients"]
```

```python
# EXPORT CSV
dim_product.to_csv("dim_product.csv", index=False)
```
